# DevPilot: Empirical Evaluation Analysis

This notebook loads empirical evaluation results from the results directory and regenerates publication figures in output_images.

If no evaluation results are present, the notebook cleanly reports that status without raising an error.


```python
import json
import os
from pathlib import Path

results_dir = Path("../results")
if not results_dir.exists():
    results_dir = Path("results")

output_images_dir = Path("../output_images")
if not output_images_dir.exists():
    output_images_dir = Path("output_images")
output_images_dir.mkdir(parents=True, exist_ok=True)

report_file = results_dir / "latest_evaluation_report.json"

if not report_file.exists():
    print("no results yet: results/latest_evaluation_report.json does not exist")
else:
    try:
        with open(report_file, "r", encoding="utf-8") as f:
            data = json.load(f)
        configs = data.get("configs", {})
        if not configs:
            print("no results yet: evaluation report contains no configuration runs")
        else:
            print("DevPilot Evaluation Report Summary")
            print("=================================")
            print(f"Total Tasks: {data.get('total_tasks', 0)}")
            print(f"Seeds Evaluated: {data.get('seeds', [])}")
            print()
            for name, cfg in configs.items():
                pass_rate = cfg.get("test_pass_rate", 0.0) * 100.0
                comp_rate = cfg.get("completion_rate", 0.0) * 100.0
                tokens = cfg.get("avg_tokens", 0.0)
                cost = cfg.get("avg_cost_usd", 0.0)
                dur = cfg.get("avg_duration_sec", 0.0)
                print(f"Configuration: {name}")
                print(f"  Pass Rate:       {pass_rate:.1f}%")
                print(f"  Completion Rate: {comp_rate:.1f}%")
                print(f"  Mean Tokens:     {tokens:.0f}")
                print(f"  Mean Cost:       ${cost:.4f}")
                print(f"  Mean Duration:   {dur:.1f}s")
                print()
    except Exception as exc:
        print(f"Error reading results: {exc}")
```
